```python
from pulp import LpMinimize, LpProblem, LpVariable

model = LpProblem("Goal_Programming", LpMinimize)

x1 = LpVariable('p1', lowBound=0)
x2 = LpVariable('p2', lowBound=0)

d1_minus = LpVariable('d1_minus', lowBound=0)
d1_plus = LpVariable('d1_plus', lowBound=0)

d2_minus = LpVariable('d2_minus', lowBound=0)
d2_plus = LpVariable('d2_plus', lowBound=0)

d3_minus = LpVariable('d3_minus', lowBound=0)
d3_plus = LpVariable('d3_plus', lowBound=0)

model += d1_minus + d2_plus + d3_plus

model += 5*x1 + 4*x2 + d1_minus - d1_plus == 100
model += 2*x1 + 1*x2 + d2_minus - d2_plus == 80
model += 1*x1 + 2*x2 + d3_minus - d3_plus == 60

model.solve()

print("jumlah produksi optimal:")
print("p1:", x1.varValue)
print("p2:", x2.varValue)

print("\nDeviasi:")
print("profit kurang:", d1_minus.varValue)
print("Tenaga kerja lebih:", d2_plus.varValue)
print("Bahan baku lebih:", d3_plus.varValue)
```
