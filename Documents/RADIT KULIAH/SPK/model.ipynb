```python
import pulp

model = pulp.LpProblem("Epsilon_consumption", pulp.LpMaximize)

x = pulp.LpVariable('x', lowBound=0)
y = pulp.LpVariable('y', lowBound=0)
  
  
model += 40*x + 30*y

model += x + 2*y <= 40
model += 3*x + y <= 60

epsilon = 50
model += 2*x + y <= epsilon

model.solve()

print("Status:", pulp.LpStatus[model.status])
print("x =", x.varValue)
print("y =", y.varValue)
print("keuntungan =", pulp.value(model.objective))
```

```python
epsilons = [40, 45, 50, 55, 60]

for eps in epsilons:
    model = pulp.LpProblem("epsilon", pulp.LpMaximize)
    
    x = pulp.LpVariable('x', lowBound=0)
    y = pulp.LpVariable('y', lowBound=0)
    
    model += 40*x + 30*y
    
    model += x + 2*y <= 40
    model += 3*x + y <= 60
    model += 2*x + y <= eps
    
    model.solve()
    
    print(f'\nepsilon = {eps}')
    print("x =", x.varValue)                                                                                                                                                 
    print("y =", y.varValue)
    print("Profit =", pulp.value(model.objective))
```

```python
# metologi utility function
import pulp

model = pulp.LpProblem("Utility_function", pulp.LpMaximize)

x = pulp.LpVariable('x', lowBound=0)
y = pulp.LpVariable('y', lowBound=0)

w1 = 0.7 # profit
w2 = 0.3 # waktu

# fungsi utilitas
model += w1*(50*x + 40*y) - w2*(3*x + 2*y)   

model += 2*x + y <= 100
model += x + y <= 80

model.solve()

print("Status:", pulp.LpStatus[model.status])
print("x =", x.varValue)
print("y =", y.varValue)
print("keuntungan =", pulp.value(model.objective))
```

```python
# analisis sensitivitas
weights = [(0.5,0.5), (0.7,0.3), (0.9,0.1)]

for w1, w2 in weights:
    model = pulp.LpProblem("Utility", pulp.LpMaximize)
    
    x = pulp.LpVariable('x', lowBound=0)
    y = pulp.LpVariable('y', lowBound=0)
    
    model += w1*(50*x + 40*y) - w2*(3*x + 2*y)
    
    model += 2*x + y <= 100
    model += x + y <= 80
    
    model.solve() 
    
    print(f"\nBobot profit: {w1}, Bobot waktu: {w2}")
    print("x =", x.varValue)
    print("y =", y.varValue)
    print("utility =", pulp.value(model.objective))
```
