```python
class MergeSortVisual:
    def __init__(self, data):
        self.data = data
        
    def merge_sort(self, arr=None, level=1):
        if arr is None:
            arr = self.data
            
        if len(arr) <= 1:
            return arr
        
        mid = len(arr) // 2
        
        left_half = arr[:mid]
        right_half = arr[mid:]
        
        print(f"\nLangkah {level}:")
        print("Array: ", arr)
        print("kiri = ", left_half)
        print("kanan = ", right_half)
        
        sorted_left = self.merge_sort(left_half, level + 1)
        sorted_right = self.merge_sort(right_half, level + 1)
        
        merged = self.merge(sorted_left, sorted_right)
        print(f"Hasil merge: {merged}")
        return merged
    
    def merge(self, left, right):
        result = []
        i = j = 0
        
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1
                
        result.extend(left[i:])
        result.extend(right[j:])
        return result
    
    def sort(self):
        print("Array awal:", self.data)
        hasil = self.merge_sort()
        print("\nHasil akhir:", hasil)
        
data = [98, 23, 45, 14, 6, 67, 33, 42]

ms = MergeSortVisual(data)
ms.sort()
        
```

```python
class QuickSortVisualizer:
    def __init__(self, data):
        self.data = data
    
    def quick_sort(self, arr=None, level=1):
        if arr is None:
            arr = self.data
            
        if len(arr) <= 1:
            return arr
        
        pivot = arr[-1]
        
        left = [x for x in arr[:-1] if x < pivot]
        right = [x for x in arr[:-1] if x >= pivot]
        
        print(f"\nLangkah {level}:")
        print("Array: ", arr)
        print("pivot = ", pivot)
        print("Elemen < pivot = ", left)
        print("Elemen >= pivot = ", right)
        
        sorted_left = self.quick_sort(left, level + 1)
        sorted_right = self.quick_sort(right, level + 1)
        
        result = sorted_left + [pivot] + sorted_right
        print("Hasil penggabungan:", result)
        return result
    
    def sort(self):
        print("Array awal:", self.data)
        hasil = self.quick_sort()
        print("\nHasil akhir:", hasil)
        
data = [7, 2, 1, 6, 8, 5, 3]

qs = QuickSortVisualizer(data)
qs.sort()
```
