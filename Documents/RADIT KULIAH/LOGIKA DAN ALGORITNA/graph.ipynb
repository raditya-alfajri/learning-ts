```python
# contoh graph
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('A', 1), ('D', 2), ('D', 5)],
    'C': [('A', 4), ('B', 2), ('D', 1)],
    'D': [('B', 5), ('C', 1)]
}

import heapq

class Dijkstra:
    def __init__(self,graph):
        self.graph = graph
        self.distances = {}
        self.previous_nodes = {}
        
    def run(self, start_node):
        # inisialisasi jarak dan previous nodes
        self.distances = {node: float('inf') for node in self.graph}
        self.distances[start_node] = 0
        self.previous_nodes = {node: None for node in self.graph}
        
        priority_queue = [(0, start_node)]
        
        while priority_queue:
            current_distance, current_node = heapq.heappop(priority_queue)
            
            if current_distance > self.distances[current_node]:
                continue
            
            for neighbor, weight in self.graph[current_node]:
                distance = current_distance + weight
                
                if distance < self.distances[neighbor]:
                    self.distances[neighbor] = distance
                    self.previous_nodes[neighbor] = current_node
                    heapq.heappush(priority_queue, (distance, neighbor))
                
    def get_shortest_path(self, end_node):
        path = []
        current = end_node

        while current is not None:
            path.append(current)
            current = self.previous_nodes[current]
        return path[::-1] 
        
    def display_all_paths(self, start_node):
        print(f"Jarak dan jalur terpendek dari simpul {start_node}:")
        for node in self.graph:
            path = self.get_shortest_path(node)
            path_str = ' -> '.join(path)
            print(f" - ke {node}: {self.distances[node]} (jalur: {path_str})")
            
            
dijkstra_solver = Dijkstra(graph)
start_node = 'D'
dijkstra_solver.run(start_node)
dijkstra_solver.display_all_paths(start_node)
```
