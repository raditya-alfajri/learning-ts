```python
class BinaryHeap:
    def __init__(self, elements):
        self.heap = [None] + elements
        
    def parent(self, i):
        return i // 2 if i > 1 else None
    
    def left_child(self, i):
        idx = 2 * i
        return idx if idx < len(self.heap) else None
    
    def right_child(self, i):
        idx = 2 * i + 1
        return idx if idx < len(self.heap) else None
    
    def get_value(self, i):
        if i is None or i >= len(self.heap):
            return None
        return self.heap[i]
    
    def print_node_info(self, i):
        if i + i < 1 or i >= len(self.heap):
            print("index tidak valid")
            return
        
        print(f"Node index yang dicari{i} (value = {self.heap[i]}):")
        print(f"Parent index {i}  {self.parent(i)} (value = {self.get_value(self.parent(i))})")
        print(f"Left Child Index {i} : {self.left_child(i)} (value = {self.get_value(self.left_child(i))})")
        print(f"Right Child Index {i} : {self.right_child(i)} (value = {self.get_value(self.right_child(i))})")
        
binary_tree = [50, 30, 40, 10, 20, 35, 25]
heap = BinaryHeap(binary_tree)
index_yang_dicari = 3
heap.print_node_info(index_yang_dicari)        
```

```python
class MaxHeap:
    def __init__(self):
        self.heap = []

    def insert(self, value):
        self.heap.append(value)
        self._heapify_up(len(self.heap) - 1)

    def build_max_heap(self, array):
        self.heap = array[:]
        for i in range(len(self.heap) // 2 - 1, -1, -1):
            self._heapify_down(i)

    def _heapify_up(self, index):
        parent = (index - 1) // 2
        if index > 0 and self.heap[index] > self.heap[parent]:
            self._swap(index, parent)
            self._heapify_up(parent)

    def _heapify_down(self, index):
        largest = index
        left = 2 * index + 1
        right = 2 * index + 2

        if left < len(self.heap) and self.heap[left] > self.heap[largest]:
            largest = left
        if right < len(self.heap) and self.heap[right] > self.heap[largest]:
            largest = right

        if largest != index:
            self._swap(index, largest)
            self._heapify_down(largest)

    def _swap(self, i, j):
        self.heap[i], self.heap[j] = self.heap[j], self.heap[i]

    def delete(self):
        if not self.heap:
            return None

        root = self.heap[0]

        if len(self.heap) == 1:
            return self.heap.pop()

        self.heap[0] = self.heap.pop()
        self._heapify_down(0)

        return root

    def get_heap(self):
        return self.heap

# Pengujian dengan array pada gambar
arr = [10, 15, 21, 30, 18, 19]
heap = MaxHeap()
heap.build_max_heap(arr)
print("Max Heap:", heap.get_heap())

tambah = 12
heap.insert(tambah)
print("Tambah data", tambah)
print("Max Heap setelah tambah data:", heap.get_heap())

deleted = heap.delete()

print("Data Dihapus :", deleted)
print("Heap Setelah Delete :", heap.get_heap())
```
