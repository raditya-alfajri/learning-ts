```python
def hitung_luas_persegi(sisi):
    luas = sisi * sisi
    print(f"luas persegi dengan sisi {sisi} adalah {luas}")
    
hitung_luas_persegi(10)
hitung_luas_persegi(5)
hitung_luas_persegi(99)
```

```python
def perkenalan(nama, umur, kota = 'Tegal' ):
    print(f"Nama saya {nama}")
    print(f"Umur saya {umur}")
    print(f"Berasal dari {kota}")
    
perkenalan("Andi", 19, 'Bandung')  
```

RETURN

```python
def tambah(a, b):
    hasil = a + b
    return hasil

#menyimpan hasil ke variabel
total = tambah (5, 3)
print(total)
```

```python
def hitung_lingkaran(jari_jari):
    phi = 3.14
    luas = phi * jari_jari * jari_jari
    keliling = 2 * phi * jari_jari
    return luas, keliling

# menerima multiple values
l, k = hitung_lingkaran(7)
print(f"Luas {l} keliling {k}")
```
