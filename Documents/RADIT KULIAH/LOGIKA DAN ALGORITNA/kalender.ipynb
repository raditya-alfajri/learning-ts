KALENDER

```python
class Kalender:
    def __init__(self, tanggal):
        self.tanggal = tanggal
        self.next = None
        
class Tanggal:
    def __init__(self):
        self.pertama = None
        
    def tampil(self):
        temp = self.pertama
        while temp:
            print(temp.tanggal, end=' => ')
            temp = temp.next
        print('None')
        
    def next(self, tanggal):
        new_tanggal = Kalender(tanggal)
        if not self.pertama:
            self.pertama = new_tanggal
        else:
            temp = self.pertama
            while temp.next:
                temp = temp.next
            temp.next = new_tanggal
        
    def delete(self, key):
        temp = self.pertama
        if temp and temp.tanggal == key:
            self.pertama = temp.next
            return
        
        while temp and temp.next:
            if temp.next.tanggal == key:
                temp.next = temp.next.next
                return
            temp = temp.next
        
sll = Tanggal()
sll.next(1)
sll.next(2)
sll.next(3)
sll.next(4)
sll.next(5)
sll.next(6)
sll.next(7)        
sll.next(8)
sll.next(9)
sll.next(10)
sll.next(11)
sll.next(12)
sll.next(13)
sll.next(14)
sll.next(15)
sll.delete(2)
sll.tampil()
```
