```python
def hitung_persegi_panjang(panjang, lebar):
    luas = panjang * lebar
    keliling = 2 * (panjang + lebar)
    return luas, keliling


l, k = hitung_persegi_panjang(10, 5)
print(f" luas {l} Keliing {k}")
```

```python
def cek_genap_ganjil(angka):
    if angka % 2 == 0:
        return "genap"
    return "ganjil"

print(cek_genap_ganjil(7))
```
