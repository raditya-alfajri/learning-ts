```python
nilai = int(input('masukkan nilai siswa: '))
print(f"nilai siswa: {nilai}")

if nilai < 0 or nilai > 100:
    print("nilai input tidak valid")
elif nilai >= 90:
    print('kategori: sangat baik')
elif nilai >= 75:
    print('kategori: baik')
elif nilai >= 60:
    print('kategori: cukup')
elif nilai >= 40:
    print('kategori: kurang')
else:
    print('kategori: sangat kurang')

        
```

```python
barang = input('masukkan status barang (biasa/premium) ') 
ukuran = input('masukkan ukuran (besar/kecil)')
jumlah = int(input('masukkan jumlah barang: '))
harga = 0 

if barang == 'biasa':
    if ukuran == 'besar':
        harga = 100000
    elif ukuran == 'kecil':
        harga = 50000
elif barang == 'premium':
    if ukuran == 'besar':
        harga = 200000
    elif ukuran == 'kecil':
        harga = 100000
else:
    print('input tidak valid')
    
diskon = 10
jumlahdiskon = (harga * jumlah * diskon)/100
total = jumlah * harga
if jumlah > 3:
    total -= jumlahdiskon
    print(f'total harga setelah didiskon: {total}')
else:
    totalharga = total
    print(f'total harga: {totalharga}')
```
