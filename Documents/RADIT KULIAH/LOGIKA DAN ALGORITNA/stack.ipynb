```python

stack = []
max = 5

def push(element):
    if not is_full():
        stack.append(element)
        print('stack sudah ditambahkan')
    else:
        stack.append(element)
        
def pop():
    stack.pop

def is_full():
    return len(stack) >= max

def peek():
    if len(stack) == 0:
        print('Stack kosong')
    else:
        return stack[-1]
    
push(1)
push(2)
push(4)
push(1)
push(1)
push(1)
push(1)
peek()
is_full()
print(stack)

```

```python
history = []

def buka_halaman(url):
    history.append(url)
    print(f"Buka halaman: {url}")

def back():
    if history:
        halaman = history.pop()
        print(f"Kembali dari: {halaman}")
    else:
        print("History kosong")

def lihat_history():
    if history:
        print("Riwayat browser:", history)
    else:       
        return back()
    

buka_halaman("google.com")
buka_halaman("youtube.com")
buka_halaman("github.com")
buka_halaman("Roblox.com")

lihat_history()
back()
back()
lihat_history()
```
