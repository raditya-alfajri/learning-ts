```python
list2 = [17, 10, 5, 6]
print(list2)

list2.append(9) # menambah indeks
print(list2)

list2.insert(3, 7) # menyisipkan indeks
print(list2)

list2[0] = 3 #mengubah indeks
print(list2)

list2.sort(reverse=True)
print(list2)

list2.remove(10)
print(list2)

list2.pop(0)
print(list2)

list2.clear()
print(list2)
```

Dictionary

```python
mahasiswa = {
    "nama": ["Raditya", "Arip"],
    "jurusan": "Teknik Informatika",
    "angkatan": 2025
}

print(mahasiswa["nama"][0], mahasiswa["jurusan"], mahasiswa["angkatan"])
kalimat = f'nama saya {mahasiswa["nama"][1]}'
print(kalimat)

mahasiswa["jurusan"] = "Keperawatan"
print(mahasiswa)

mahasiswa["nama"][1] = "dapa"
print(mahasiswa)

mahasiswa["kelas"] = "A"
print(mahasiswa)

mahasiswa["nama"].append("Pani")
print(mahasiswa)

mahasiswa["nama"].pop(2)
print(mahasiswa)
```

Studi Kasus

```python
siswa = {
    'Ali': 95,
    'Radit': 100,
    'Arip': 90,
}

def cek_nilai(siswa, nama):
    if nama in siswa:
        return siswa[nama]
    else:
        return f'Data siswa dengan nama {nama} tidak ditemukan'
    
hasil = cek_nilai(siswa, 'Dika')
print(hasil)
```
