```python
from collections import deque

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None        
        
class BinaryTree:
    def __init__(self):
        self.root = None
        
    def insert(self, data):
        new_node = Node(data)
        
        if self.root is None:
            self.root = new_node
            return
        
        queue = deque([self.root])
        
        while queue:
            current = queue.popleft()
            
            if current.left is None:
                current.left = new_node
                return
            else:
                queue.append(current.left)
                
            if current.right is None:
                current.right = new_node
                return
            else:
                queue.append(current.right)
            
    def search(self, target):
        if self.root is None:
            return False
        
        queue = deque([self.root])
        
        while queue:
            current = queue.popleft()
            
            if current.data == target:
                return True
            
            if current.left:
                queue.append(current.left)
                
            if current.right:
                queue.append(current.right)
                
        return False
    
    
    # Delete
    def delete(self, target):
        if self.root is None:
            return

        if self.root.left is None and self.root.right is None:
            if self.root.data == target:
                self.root = None
            return

        queue = deque([self.root])

        target_node = None
        last_node = None
        parent_last = None

        while queue:
            current = queue.popleft()

            # cari node target
            if current.data == target:
                target_node = current

            if current.left:
                parent_last = current
                queue.append(current.left)

            if current.right:
                parent_last = current
                queue.append(current.right)

            last_node = current

        # kalau target ketemu
        if target_node:
            # ganti data target dengan deepest node
            target_node.data = last_node.data

            # hapus deepest node
            if parent_last.right == last_node:
                parent_last.right = None
            else:
                parent_last.left = None
    
    
    # in-order-DFS: left, Root, Right
    def in_order_dfs(self, node):
        if node is None:
            return
        self.in_order_dfs(node.left)
        print(node.data, end=" ")
        self.in_order_dfs(node.right)
    
    # pre-order-DFS: root, left, right    
    def pre_order_dfs(self, node):
        if node is None:
            return
        print(node.data, end=' ')
        self.pre_order_dfs(node.left)
        self.pre_order_dfs(node.right)
        
    # pre-order-DFS: root, left, right    
    def post_order_dfs(self, node):
        if node is None:
            return
        self.post_order_dfs(node.left)
        self.post_order_dfs(node.right)
        print(node.data, end=' ')



bt = BinaryTree()

bt.insert(2)
bt.insert(3)
bt.insert(4)
bt.insert(5)
bt.insert(6)

print("in_order DFS: ", end='')
bt.in_order_dfs(bt.root)
print("\npre_order DFS: ", end='')
bt.pre_order_dfs(bt.root)
print("\npost_order DFS: ", end='')
bt.post_order_dfs(bt.root)

print("\nIn Order Sebelum delete:")
bt.in_order_dfs(bt.root)

bt.delete(6)

print("\n\nIn Order Sesudah delete 6:")
bt.in_order_dfs(bt.root)


print("\n\nCari 5:", bt.search(5))
print("Cari 10:", bt.search(10))
```
