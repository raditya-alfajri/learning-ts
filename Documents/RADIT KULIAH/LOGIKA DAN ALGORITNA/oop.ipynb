```python
class Kendaraan:
    def __init__(self, merk, model, harga):
        self.merk = merk
        self.model = model 
        self.harga = harga
        
    def info(self):
        return f"{self.merk} {self.model}"   
    
    def info_harga(self):
        output = f"Harga Kendaraan dengan merk {self.merk} adalah {self.harga}"
        return output
        
mobil = Kendaraan("Toyota", "Avanza", 120000000)
print(mobil.info())
print(mobil.info_harga())

        
        
```

```python
class Mahasiswa:
    def __init__(self, nama, nim, jurusan):
        self.nama = nama
        self.nim = nim
        self.jurusan = jurusan
        
    
    def info_mahasiswa(self):
        output = f"Perkenalkna saya {self.nama} dengan nim {self.nim} dan saya dari jurusan {self.jurusan}"
        return output
    
mahasiswa = Mahasiswa("Radit", "25090008", "Teknik Informatika")
print(mahasiswa.info_mahasiswa())
```
