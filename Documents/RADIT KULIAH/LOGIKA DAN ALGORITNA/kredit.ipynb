```python
import matplotlib.pyplot as plt
import networkx as nx

# 1. Membuat objek graf
G = nx.Graph()

# 2. Menambahkan edge (hubungan parent-child)
# Contoh hierarki: Root -> A -> B
edges = [('Root', 'A'), ('Root', 'B'), 
         ('A', 'A1'), ('A', 'A2'), 
         ('B', 'B1'), ('B', 'B2')]
G.add_edges_from(edges)

# 3. Mengatur tata letak (Layout) agar berbentuk pohon
# Menggunakan 'shell_layout' atau 'spring_layout' sebagai alternatif
pos = nx.spring_layout(G) 

# 4. Menggambar grafik menggunakan matplotlib
plt.figure(figsize=(8, 6))
nx.draw(G, pos, with_labels=True, node_size=2000, 
        node_color='skyblue', font_size=10, 
        font_weight='bold', edge_color='gray')

# 5. Menampilkan hasil
plt.title("Contoh Grafik Tree di Python")
plt.show()
```
