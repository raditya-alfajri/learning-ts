1.C (Circular Linked List)

```python
class JamDinding:
    def __init__(self, jam):
        self.jam = jam
        self.next = None
        
class CircularLinkedList:
    def __init__(self):
        self.head = None

    def tampil(self):
        if self.head is None:
            print("Jam Kosong")
            return
        
        temp = self.head
        
        while True:
            print(temp.jam, end=' -> ')
            temp = temp.next
            if temp == self.head:
                break
            
    def jam_akhir(self, jam):
        new_angka = JamDinding(jam)
        if self.head is None:
            self.head = new_angka
            new_angka.next = self.head
            return
        
        temp = self.head
        
        while temp.next != self.head:
            temp = temp.next
            
        temp.next = new_angka
        new_angka.next = self.head
        
    def jam_awal(self, jam):
        new_angka = JamDinding(jam)
        if self.head is None:
            self.head = new_angka
            new_angka.next = self.head
            return
        
        temp = self.head
        
        while temp.next != self.head:
            temp = temp.next
            
        new_angka.next = self.head
        temp.next = new_angka
        self.head = new_angka    
        
        print()
        
cll = CircularLinkedList()
cll.jam_akhir(7)
cll.jam_akhir(8)
cll.jam_akhir(9)
cll.jam_akhir(10)
cll.jam_akhir(11)
cll.jam_akhir(12)
cll.jam_awal(6)
cll.tampil() 
        
```

2. A (Merge Sort)

```python
class NilaiMahasiswa:
    def __init__(self, nilai):
        self.nilai = nilai
        
    def merge_sort(self, array=None, level=1):
        if array is None:
            array = self.nilai
            
        if len(array) <= 1:
            return array
        
        mid = len(array) // 2
        
        left_half = array[:mid]
        right_half = array[mid:]
        
        print(f"\nLangkah {level}:")
        print("Array: ", array)
        print("kiri = ", left_half)
        print("kanan = ", right_half)
        
        sorted_left = self.merge_sort(left_half, level + 1)
        sorted_right = self.merge_sort(right_half, level + 1)
        
        merged = self.gabung(sorted_left, sorted_right)
        print(f"Hasil merge: {merged}")
        return merged
    
    def gabung(self, left, right):
        result = []
        i = j = 0
        
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1
                
        result.extend(left[i:])
        result.extend(right[j:])
        return result
    
    def sort(self):
        print("Array awal:", self.nilai)
        hasil = self.merge_sort()
        print("\nHasil akhir:", hasil)
        
data = [98, 100, 70, 80, 85, 67, 75, 60, 77]

ms = NilaiMahasiswa(data)
ms.sort()
        
```

cara kerja merge sort adalah 

langkah pertama:
ditampilkan array pertama dulu yang masih belum terurut

langkah kedua:
data akan dicari tengah2 nya dan dibagi dua itu untuk masing2 bagian diurutkan

langkah ketiga:
setelah tadi data sudah dibagu menajadi dua masing2 data akan di bagi lagi hingga masing2 menjadi elemen satu

langkah keempat:
jika sudah terbagi menjadi elemen satu makan akan digabungkan lagi menjadi dua bagian pada sisi kiri dan sisi kanan dengan posisi data teurut masing2 bagian

langkah kelima:
dua bagian data tadi yang sudah digabungkan dan urut makan dua bagian tadi digabungkan menjadi satu dengan kondisi data terurut

itulah cara kerja merge sort yang akhirnya setelah dibagi dan diurutkan menjadi satu utuh data yang terurut

```python
class BinarySearch:
    def __init__(self, array):
        self.array = sorted(array)



    def search(self, target):
        low = 0
        high = len(self.array) - 1
        
        while low <= high:
            mid = low + (high - low) // 2
            
            if self.array[mid] == target:
                return mid
            elif self.array[mid] < target:
                low = mid + 1
            else:
                high = mid - 1
        
        return -1

    def display_result(self, target):
        result = self.search(target)  
        if result != -1:
            print(f"elemen {target} ditemukan di indeks {result}")
        else:
            print(f"elemen {target} tidak ditemukan dalam array")
            
data = [34, 7, 23, 32, 5, 62, 78, 1, 45, 19]
target = 5


searcher = BinarySearch(data)
searcher.display_result(target)      
            
        
```

ALUR PROGRAM BINERY SEARCH

binary search bekerja dengan cara membagi ruang pencarian menjadi dua. Proses ini dilakukan dengan memeriksa elemen tengah pada data yang sudah terurut, lalu terus mengeliminasi setengah bagian yang tidak mungkin memuat data target hingga nilai yang dicari ditemukan

PENJELASAN LOW, MID, HIGH

Kondisi 1 (Data Ditemukan): Jika nilai pada indeks mid sama dengan nilai target yang dicari, program berhenti dan mengembalikan posisi indeks tersebut.Kondisi 2 (Nilai Terlalu Kecil): Jika nilai target lebih besar dari elemen tengah (mid), berarti target berada di bagian kanan. 
Kondisi 3 (Nilai Terlalu Besar): Jika nilai target lebih kecil dari elemen tengah (mid), berarti target berada di bagian kiri.
jika perulangan selesai dan kondisi LOW, MID, HIGH tidak lagi terpenuhi, maka data target dipastikan tidak ada di dalam kumpulan data.
