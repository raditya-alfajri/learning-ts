LINEAR SEARCH

```python
class LinearSearch:
    def __init__(self, array):
        self.array = array
        
    def search(self, target):
        for index, value in enumerate(self.array):
            if value == target:
                return index
        return -1
    
    def display_result(self, target):
        result = self.search(target)
        if result != -1:
            print(f"elemen {target} ditemukan di indeks {result}")
        else:
            print(f"elemen {target} tidak ditemukan dalam array.")
            
data = [2, 4, 6, 8, 10]
target = 8
searcher = LinearSearch(data)
searcher.display_result(target)
            
```

BINARY SEARCH

```python
class BinarySearch:
    def __init__(self, array):
        self.array = sorted(array)



    def search(self, target):
        low = 0
        high = len(self.array) - 1
        
        while low <= high:
            mid = low + (high - low) // 2
            
            if self.array[mid] == target:
                return mid
            elif self.array[mid] < target:
                low = mid + 1
            else:
                high = mid - 1
        
        return -1

    def display_result(self, target):
        result = self.search(target)  
        if result != -1:
            print(f"elemen {target} ditemukan di indeks {result}")
        else:
            print(f"elemen {target} tidak ditemukan dalam array")
            
data = [3, 1, 7, 5, 9, 2]
target = 7


searcher = BinarySearch(data)
searcher.display_result(target)      
            
        
```
