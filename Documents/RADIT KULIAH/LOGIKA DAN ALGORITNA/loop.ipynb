```python
listkota = [
    'jogjakarta', 'jakarta', 'bandung', 'tegal', 'semarang',
    'bogor', 'bekasi'
]

kotayangdicari = input('kota yang anda cari: ')
print(f'kota yang anda cari: {kotayangdicari}')

for i, kota in enumerate(listkota):
    if kota.lower() == kotayangdicari.lower():
        print('kota yang anda cari berada di indeks: ', i)
        break
else:
    print('maaf kota yang anda cari tidak ada')
```
