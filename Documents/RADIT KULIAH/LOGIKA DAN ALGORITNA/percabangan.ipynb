IF

```python
if True:
    print('kode ini akan tercetak')
if False:
    print('kode ini tidak akan tercetak')
    
print('kode ini akan selalu tercetak')
```

```python
if 5 > 10:
    print('5 lebih dari 10')
if 10 > 5:
    print('10 lebih dari 5')
    
```

if else

```python
nilai = int(input('masukkan nilai anda: '))
print(f'nilai anda: {nilai}')

if nilai >= 70:
    print('lulus')
else:
    print('tidak lulus')
```

if else elif

```python
nilai = int(input('masukkan nilai :'))
print(f'nilai anda {nilai}')

if nilai >= 90:
    print('predikat A')
elif nilai >= 80:
    print('predikat B')
elif nilai >= 60:
    print('predikat c')
elif nilai >= 40:
    print('predikat D')
else:
    print('predikat E')
```

boolean

```python
buah_yang_tersedia = ('jeruk', 'mangga', 'melon')
buah_yang_dicari = input('masukkan buah yang dicari: ').lower()
print(f'buah yang dicari {buah_yang_dicari}')

if (buah_yang_dicari in buah_yang_tersedia):
    print('buah yang anda cari tersedia')
else:
    print('buah yang anda cari tidak tersedia')
```

cabang satu baris

```python
nilai = int(input('masukkan nilai: '))
print(f'nilai: {nilai}')
status = 'lulus' if nilai >= 70 else 'tidak lulus'

print(status)
```

cabang bertingkat

```python
nilai = int(input('masukkan nilai anda: '))
usia = int(input('masukkan usia anda: '))

if nilai >= 75:
    if (usia < 15):
        print('selamat adek, kamu lulus')
    else:
        print('selamat kakak, kamu lulus')
else:
    if (usia < 15):
        print('mohon maaf dek, coba lagi ya')
    else:
        print('mohon maaf kak, coba lagi ya')
```
