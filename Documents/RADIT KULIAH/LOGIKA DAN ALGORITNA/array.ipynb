```python
list1 = ['ikan', 'kucing', 'beruang']
list1 = np.array(list1)
print(list1)
print(list1.shape)
print('ini adalah array', list1.ndim ,'D')
```

ARRAY 2D

```python
array_2d = np.array(
    [
        [10, 20, 30],
        [40, 50, 60],
        [70, 80, 90]
    ]
)

print(array_2d)
print(array_2d.shape)
print('ini array', array_2d.ndim, 'D')
```

```python
array_2d_string = np.array(
    [
        ['apple', 'banana', 'cherry'],
        ['dog', 'elephant', 'fox'],
        ['grape', 'honey', 'ice'],
        
    ]
)
print(array_2d_string[1, 1])
array_2d_string[1, 1] = 'cat'
print(array_2d_string)
```
