```python
class BubbleShort:
    def __init__(self, data):
        self.data = data
        
    def sortir(self):
        n = len(self.data)
        for i in range(n):
            for j in range(0, n - i - 1):
                if self.data[j] > self.data[j + 1]:
                   self.data[j], self.data[j + 1] = self.data[j + 1], self.data[j]
            print(f"Langkah {i+1}: {self.data}")
            
angka = [75, 60, 90, 55, 80, 70]
proses = BubbleShort(angka)

print("Sebelum diurutkan", angka)
proses.sortir()
print("sesudah diurutkan", angka)
                    
```

```python
class InsertionSort:
    def __init__(self, data):
        self.data = data
        
    def sortir(self):
        for i in range(1, len(self.data)):
            key = self.data[i]
            j = i - 1
            
            while j >= 0 and self.data[j] > key:
                self.data[j + 1] = self.data[j]
                j -= 1
            self.data[j + 1] = key
            print(f"Langkah {i}: {self.data}")
        
nilai = [75, 60, 90, 55, 80, 70]
proses = InsertionSort(nilai)

print("sebelum diurutkan", nilai)
proses.sortir()
print("sesudah diurutkan",)
```

```python
class SelectionSort:
    def __init__(self, data):
        self.data = data
        
    def sortir(self):
        n = len(self.data)
        for i in range(n):
            indeks_terkecil = i
            for j in range(i + 1, n):
                if self.data[j] < self.data[indeks_terkecil]:
                    indeks_terkecil = j
                    
            self.data[i], self.data[indeks_terkecil] = self.data[indeks_terkecil], self.data[i]
            print(f"Langkah {i+1}: {self.data}")
            
nilai = [75, 60, 90, 55, 80, 70]
proses = SelectionSort(nilai)

print("sebelum diurutkan", nilai)
proses.sortir()
print("sesudah diurutkan", nilai)                                                      
```
