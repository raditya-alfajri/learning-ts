```python
nilai = int(input('masukkan nilai siswa: '))
print(f'nilai: {nilai}')

if nilai < 0 or nilai > 100:
    print('nilai tidak valid')
elif nilai >= 85:
    print('A (sangat Baik)')
elif nilai >= 75:
    print('B (Baik)')
elif nilai >= 65:
    print('C (Cukup)')
elif nilai >= 50:
    print('D (Kurang)')
else:
    print('E (Gagal)')    
```

```python
karyawan = []

for i in range(3):  
    nama = input("Nama karyawan: ")
    gaji_pokok = int(input("Gaji pokok: "))
    lama_kerja = int(input("lama kerja: "))

    karyawan.append([nama, gaji_pokok, lama_kerja])
    
print("\n=== HASIL PERHITUNGAN GAJI ===")

for data in karyawan:
    nama, gaji, lamakerja = data

    if lamakerja >= 5:
        persen = 15
    elif lamakerja >= 3:
        persen = 10
    else:
        persen = 8

    bonus = gaji * persen / 100
    total_gaji = gaji + bonus

    print(f"nama karyawan: {nama}")
    print(f"Gaji Pokok: {gaji}")
    print(f"Bonus: {bonus}")
    print(f"Total: {total_gaji}")
    print()
    
    
```
